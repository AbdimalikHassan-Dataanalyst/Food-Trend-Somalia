```python

# 1. Hubi records-ka USD price ka maqan yahay
missing_usd = df[df["usdprice"].isna()]

print("Missing USD price records:", len(missing_usd))

# 2. Eeg qiimaha asalka ah iyo lacagta lagu qoray
print(
    missing_usd[
        ["date", "admin1", "market", "commodity",
         "unit", "currency", "price", "usdprice"]
    ].head(15).to_string(index=False)
)

# 3. Hubi currency-yada records-kaas
print("\nCurrency distribution:")
print(missing_usd["currency"].value_counts(dropna=False))

# 4. Hubi in price asalka ahi isna maqan yahay
print("\nMissing original prices:")
print(missing_usd["price"].isna().sum())

# 5. Hubi qiimaha USD ee maqan, sannad kasta
missing_usd["year"] = pd.to_datetime(
    missing_usd["date"]
).dt.year

print("\nMissing USD prices by year:")
print(missing_usd.groupby("year").size())
```

```python
# Change "usdprice" to your actual price column
price_col = "usdprice"

df[price_col] = pd.to_numeric(
    df[price_col], errors="coerce"
)

price = df[price_col].dropna()

overall_stats = pd.Series({
    "Count": price.count(),
    "Mean": price.mean(),
    "Median": price.median(),
    "Minimum": price.min(),
    "Maximum": price.max(),
    "Standard_Deviation": price.std(),
    "Variance": price.var(),
    "Q1": price.quantile(0.25),
    "Q3": price.quantile(0.75),
    "IQR": price.quantile(0.75) - price.quantile(0.25),
    "Coefficient_of_Variation_%": (
        price.std() / price.mean() * 100
        if price.mean() != 0 else np.nan
    )
})

display(overall_stats.to_frame("Value"))
```
